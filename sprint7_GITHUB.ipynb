**Interpretació:** aquesta funció fa de formulari. Recull totes les dades d'un sol candidat i les agrupa en un diccionari perquè la resta de funcions puguin treballar amb una única estructura.

🧩 **Desgranant el problema: naixement de les funcions**

✔ `obtenir_dades()`

1. Crear un diccionari buit per guardar les dades d'un candidat.
2. Demanar el nom i cognoms via `input` i guardar-los al diccionari.
3. Demanar el telèfon de contacte via `input` i guardar-lo al diccionari.
4. Demanar els anys d'experiència via `input` i guardar-los al diccionari.
5. Demanar el nivell de català, mostrant les opcions de 0 a 3, i guardar la resposta al diccionari.
6. Demanar el nivell d'Excel, mostrant les opcions de 0 a 3, i guardar la resposta al diccionari.
7. Demanar el nivell de titulació, mostrant les opcions de 0 a 3, i guardar la resposta al diccionari.
8. Retornar el diccionari amb totes les dades del candidat.


```python
def obtenir_dades():

    dicc = {}

    dicc["nom"] = input("Escriu el teu nom i cognoms")
    dicc["telf"] = input("Escriu el teu telèfon de contacte")
    dicc["exp"] = input("Quants anys d'experiència tens dins l'empresa?")
    dicc["cat"] = input(
        "Coneixement de català:\n"
        "0 = Sense coneixement\n"
        "1 = Fins a nivell B1\n"
        "2 = Nivell B2\n"
        "3 = C1 o superior\n"
        "Escull una opció (0-3): ")
    dicc["excel"] = input(
        "Domini d'Excel:\n"
        "0 = nivell bàsic\n"
        "1 = ús de funcions\n"
        "2 = taules dinàmiques\n"
        "3 = Visual Basic i Macros\n"
        "Escull una opció (0-3): ")
    dicc["titol"] = input(
        "Títol acadèmic:\n"
        "0 = ESO\n"
        "1 = FP\n"
        "2 = Universitat\n"
        "3 = Màster o superior\n"
        "Escull una opció (0-3): ")

    return dicc
```

**Interpretació:** aquesta funció comprova si les dades numèriques es poden utilitzar correctament. Si es poden convertir a enters retorna `True`; si alguna conversió falla, retorna `False`.

✔ `validar_dades(p_dicc)`

1. Rebre com a paràmetre el diccionari d'un candidat.
2. Tractar de convertir a números enters les dades que després s'utilitzaran en els càlculs:
   - experiència,
   - català,
   - Excel,
   - titulació.
3. Si totes les conversions es poden fer correctament, retornar `True`.
4. Si alguna conversió provoca un `ValueError`, mostrar un missatge d'error i retornar `False`.


```python
def validar_dades(p_dicc):

    try:
        p_dicc["exp"] = int(p_dicc["exp"])
        p_dicc["cat"] = int(p_dicc["cat"])
        p_dicc["excel"] = int(p_dicc["excel"])
        p_dicc["titol"] = int(p_dicc["titol"])

        if p_dicc["exp"] < 0:
            return False
        if p_dicc["cat"] <0 or p_dicc["cat"] >3:
            return False
        if p_dicc["excel"] <0 or p_dicc["excel"] >3:
            return False
        if p_dicc["titol"] <0 or p_dicc["titol"] >3:
            return False

        return True

    except ValueError:
        print("Error: Si us plau, introdueix números vàlids.")
        return False
```

**Interpretació:** aquesta funció transforma les dades del candidat en una puntuació numèrica aplicant la fórmula indicada a l'enunciat.

✔ `calcular_puntuacio(p_dicc)`

1. Rebre com a paràmetre el diccionari del candidat ja validat.
2. Multiplicar els anys d'experiència per `1.5`.
3. Sumar al resultat els punts de català.
4. Sumar els punts d'Excel.
5. Sumar els punts de titulació.
6. Guardar el resultat en la puntuació del candidat.
7. Retornar la puntuació calculada.


```python
def calcular_puntuacio(p_dicc):

    return (p_dicc["exp"] * 1.5) + p_dicc["cat"] + p_dicc["excel"] + p_dicc["titol"]
```

**Interpretació:** aquesta funció rep una puntuació i decideix a quin dels tres grups de prioritat pertany el candidat.

✔ `classificar_candidat(p_puntuacio)`

1. Rebre com a paràmetre la puntuació del candidat.
2. Si la puntuació és igual o superior a 12, assignar la classificació `Prioritat Alta`.
3. Si no, però la puntuació és igual o superior a 8, assignar la classificació `Candidat adequat`.
4. Si no es compleix cap de les condicions anteriors, assignar la classificació `No prioritari`.
5. Retornar la classificació obtinguda.


```python
def classificar_candidat(p_puntuacio):
    if p_puntuacio >= 12:
        classificacio = "Prioritat Alta"
    elif p_puntuacio >= 8:
        classificacio = "Candidat adequat"
    else:
        classificacio = "No prioritari"
    return classificacio
```

**Interpretació:** aquesta funció només s'encarrega de mostrar a cada candidat la puntuació que ha obtingut després del càlcul.

✔ `mostrar_individual(p_puntuacio_candidat)`

1. Rebre com a paràmetre la puntuació del candidat.
2. Mostrar en pantalla la puntuació individual del candidat.


```python
def mostrar_individual(p_puntuacio_candidat):
    print(f"La teva puntuació és: {p_puntuacio_candidat}")
```

**Interpretació:** aquesta funció presenta en pantalla la informació necessària per a l'informe. En la versió final haurà de mostrar els candidats ja preparats i ordenats.

✔ `mostrar_informe(p_dicc)`

1. Rebre les dades del candidat que s'han de mostrar.
2. Accedir al nom del candidat.
3. Accedir al telèfon del candidat.
4. Accedir a la puntuació del candidat.
5. Mostrar en pantalla el nom, el telèfon i la puntuació amb salts de línia per facilitar-ne la lectura.

> **Nota:** quan es construeixi el programa complet amb els 10 candidats, l'informe final haurà de treballar amb tots els candidats i mostrar-los en l'ordre demanat per l'enunciat.


```python
def mostrar_informe(p_llista_candidats):

    for candidat in p_llista_candidats:
        print(
        f"Nom: {candidat['nom']}\n"
        f"Telèfon: {candidat['telf']}\n"
        f"Puntuació: {candidat['puntuacio']}"
    )
```

**Interpretació:** aquesta és la funció que coordina tot el programa. No fa els càlculs directament, sinó que va cridant les altres funcions en l'ordre correcte fins a aconseguir 10 candidats vàlids i generar l'informe final.

🧠 **Algoritme principal**

✔ `main_seleccio()`

1. Crear una estructura per guardar els candidats vàlids.
2. Repetir el procés mentre encara no hi hagi 10 candidats vàlids.
3. Cridar `obtenir_dades()` per obtenir les dades d'un candidat.
4. Cridar `validar_dades()` per comprovar les dades introduïdes.
5. Si les dades són vàlides:
   - calcular la puntuació amb `calcular_puntuacio()`,
   - guardar la puntuació amb les dades del candidat,
   - classificar el candidat amb `classificar_candidat()`,
   - guardar la classificació amb les dades del candidat,
   - mostrar la puntuació amb `mostrar_individual()`,
   - guardar el candidat a l'estructura de candidats vàlids.
6. Si les dades no són vàlides, no comptar el candidat dins dels 10 registres vàlids.
7. Quan hi hagi exactament 10 candidats vàlids, deixar de demanar noves inscripcions.
8. Ordenar els candidats segons la prioritat indicada a l'enunciat i, en cas d'empat, segons la puntuació.
9. Mostrar l'informe final amb `mostrar_informe()`.


```python
def main_seleccio():
    llista_candidats = []

    while len(llista_candidats) < 10:
        dicc_candidat = obtenir_dades()
        validacio = validar_dades(dicc_candidat)

        if validacio is True:
            dicc_candidat["puntuacio"] = calcular_puntuacio(dicc_candidat)
            dicc_candidat["classificacio"] = classificar_candidat(dicc_candidat["puntuacio"])
            mostrar_individual(dicc_candidat["puntuacio"])
            llista_candidats.append(dicc_candidat)

    llista_candidats.sort(key=lambda candidat: candidat["puntuacio"], reverse=True)
    mostrar_informe(llista_candidats)
```

**Idea clau:** el `for` permet que `client["codi_comportament"]` faci referència al codi del client que s'està processant en aquella volta.


```python
def calcular_risc(p_clients, p_dicc_codi_comp, p_dicc_edat):

    for client in p_clients:
        client["segment"] = p_dicc_codi_comp[client["codi_comportament"]]["segment"]
        client["missatge"] = p_dicc_codi_comp[client["codi_comportament"]]["missatge"]
        client["factor_codi"] = p_dicc_codi_comp[client["codi_comportament"]]["factor_codi"]
        client["factor_edat"] = p_dicc_edat[client["grup_edat"]]
        client["risc_final"] = client["factor_codi"] * client["factor_edat"]

    return p_clients
```

### ✔ `netejar_dades(p_pacients)`

**Interpretació:** aquesta funció neteja els registres en brut i normalitza els símptomes perquè tinguin el mateix format que les claus del diccionari `nivells_risc`.

1. Rebre la llista de pacients.
2. Crear una llista buida `pacients_nets`.
3. Recórrer cada registre amb un `for`.
4. Convertir el text a minúscules amb `.lower()`.
5. Separar nom, edat i símptomes amb `.split("|")`.
6. Separar els símptomes amb `.split(",")`.
7. Eliminar espais sobrants amb `.strip()`.
8. Substituir espais interns per `_`.
9. Guardar la llista de símptomes normalitzats dins del pacient.
10. Afegir el pacient net a `pacients_nets`.
11. Retornar la llista completa.

**Idea clau:** la normalització permet que, per exemple, `Fiebre Alta` es converteixi en `fiebre_alta`.


```python
def netejar_dades(p_pacients):

    pacients_nets = []

    for pacient in p_pacients:

        pacient_net = pacient.lower()
        pacient_net = pacient_net.split("|")
        simptomes = pacient_net[2].split(",")
        simptomes_nets = []

        for simptoma in simptomes:
            simptoma = simptoma.strip()
            simptoma = simptoma.replace(" ", "_")
            simptomes_nets.append(simptoma)

        pacient_net[2] = simptomes_nets
        pacients_nets.append(pacient_net)

    return pacients_nets

dades_netes = netejar_dades(pacients)
print(dades_netes)
```

### ✔ `validar_dades(p_pacients)`

**Interpretació:** aquesta funció comprova que l'edat dels pacients es pugui convertir correctament a número abans de continuar amb el procés.

```python
def validar_dades(p_pacients):

    try:
        for pacient in p_pacients:
            pacient[1] = int(pacient[1])

        return True

    except ValueError:
        return False

dades_validades = validar_dades(dades_netes)
print(dades_netes)
```

### ✔ `calcular_risc_pacients(p_nivells_risc, p_pacients)`

**Interpretació:** aquesta funció calcula el risc total de cada pacient a partir dels símptomes normalitzats i del diccionari `nivells_risc`.

1. Rebre:

   - la llista de diccionaris dels pacients.
2. Recórrer cada pacient.
3. Inicialitzar `risc_total` a `0` per a cada pacient.
4. Recórrer la llista de `simptomes` del pacient.
5. Consultar al diccionari `p_nivells_risc` el valor de risc corresponent a cada símptoma.
6. Sumar cada valor a `risc_total`.
7. Afegir la nova clau `risc_total` al diccionari del pacient.
8. Retornar la llista de pacients actualitzada.

**Idea clau:** cada símptoma ja està normalitzat amb el mateix format que les claus de `nivells_risc`, per això es pot consultar directament el seu valor.


```python
def calcular_risc_pacients(p_nivells_risc, p_pacients):

    for pacient in p_pacients:
        risc_total = 0

        for simptoma in pacient["simptomes"]:
            simptoma = p_nivells_risc[simptoma]
            risc_total = risc_total + simptoma

        pacient["risc_total"] = risc_total

    return p_pacients

dicc_pacients_final = calcular_risc_pacients(nivells_risc, dicc_pacients)
print(dicc_pacients_final)
```

### ✔ `preparar_dades(p_df)`

**Interpretació:** aquesta funció prepara les dades originals perquè després puguem realitzar els càlculs dels partits. Assigna noms a les columnes, separa el resultat del partit en gols locals i visitants i converteix aquests valors a nombres enters.

6. Converteix les dues columnes de gols a tipus `int` mitjançant `.astype(int)`.

**Idea clau:** per calcular els gols en contra hem de mirar els gols marcats per l'equip contrari. La columna que utilitzem depèn de si l'equip està jugant com a local o com a visitant.

### Creació de `df_estadistiques`

**Interpretació:** després de calcular les diferents estadístiques, les agrupem en un únic DataFrame. A `df_partits_net` cada fila representa un partit, mentre que a `df_estadistiques` cada fila representa un equip.

5. Afegim la columna `empatats` amb els empats totals.

**Idea clau:** podem construir un DataFrame afegint Series com a columnes. Pandas les alinea segons el seu índex, de manera que les estadístiques corresponents al mateix equip queden a la mateixa fila.


### Càlcul dels punts

**Interpretació:** una vegada tenim les victòries i els empats de cada equip, podem calcular directament els punts de la classificació i guardar-los en una nova columna.

1. Multipliquem el nombre de partits guanyats per `3`.
2. Cada empat aporta `1` punt, per tant sumem directament la columna `empatats`.
3. Les derrotes aporten `0` punts i no és necessari incloure-les en el càlcul.
4. Guardem el resultat en una nova columna anomenada `punts`.
5. Retornem `df_estadistiques` amb totes les estadístiques calculades.

**Fórmula:**

`punts = (guanyats × 3) + empatats`

**Idea clau:** Pandas permet realitzar operacions matemàtiques directament entre columnes. El càlcul s'aplica automàticament a totes les files, per tant no és necessari utilitzar un bucle ni `.apply()`.

```python
def estadistics(p_df):

    # GOLS A FAVOR
    gols_local_favor = p_df.groupby("equip_local")["gols_local"].sum()
    gols_visitant_favor = p_df.groupby("equip_visitant")["gols_visitant"].sum()
    gols_totals_favor = gols_local_favor + gols_visitant_favor

    # GOLS EN CONTRA
    gols_local_contra = p_df.groupby("equip_local")["gols_visitant"].sum()
    gols_visitant_contra = p_df.groupby("equip_visitant")["gols_local"].sum()
    gols_totals_contra = gols_local_contra + gols_visitant_contra

    # GUANYATS
    guanyats_local = p_df[p_df["gols_local"] > p_df["gols_visitant"]]
    total_guanyats_local = guanyats_local.groupby("equip_local").size()
    guanyats_visitant = p_df[p_df["gols_local"] < p_df["gols_visitant"]]
    total_guanyats_visitant = guanyats_visitant.groupby("equip_visitant").size()
    total_guanyats = total_guanyats_local + total_guanyats_visitant

    # PERDUTS
    perduts_local = p_df[p_df["gols_local"] < p_df["gols_visitant"]]
    total_perduts_local = perduts_local.groupby("equip_local").size()
    perduts_visitant = p_df[p_df["gols_local"] > p_df["gols_visitant"]]
    total_perduts_visitant = perduts_visitant.groupby("equip_visitant").size()
    total_perduts = total_perduts_local + total_perduts_visitant

    # EMPATS
    empats = p_df[p_df["gols_local"] == p_df["gols_visitant"]]
    total_empats_local = empats.groupby("equip_local").size()
    total_empats_visitant = empats.groupby("equip_visitant").size()
    total_empats = total_empats_local + total_empats_visitant

    # CREAR DF
    df_estadistiques = pd.DataFrame()
    df_estadistiques["gols_favor"] = gols_totals_favor
    df_estadistiques["gols_contra"] = gols_totals_contra
    df_estadistiques["guanyats"] = total_guanyats
    df_estadistiques["empatats"] = total_empats
    df_estadistiques["perduts"] = total_perduts

    # PUNTS
    df_estadistiques["punts"] = (df_estadistiques["guanyats"] * 3) + (df_estadistiques["empatats"])

    # CANVIAR NOM DE L'ÍNDEX
    df_estadistiques.index.name = "equip"

    return df_estadistiques

df_estadistiques = estadistics(df_partits_net)
display(df_estadistiques)
```

### Classificació final

**Interpretació:** preparem la classificació final a partir del DataFrame que ja està ordenat per punts.

1. La funció rep el DataFrame d'estadístiques ordenat de major a menor puntuació.
2. En aquest DataFrame, els noms dels equips es troben a l'índex.
3. Utilitzem `.reset_index()` per convertir l'índex `equip` en una columna normal.
4. En fer `reset_index()`, Pandas crea automàticament un nou índex numèric que comença per `0`.
5. Seleccionem únicament les columnes `equip` i `punts`, que són les dades necessàries per mostrar la classificació.
6. Sumem `1` al nou índex perquè les posicions comencin per `1` en lloc de `0`.
7. Mostrem el DataFrame final.

**Idea clau:** `.reset_index()` converteix l'índex actual en una columna i crea un nou índex numèric.

En aquest cas aprofitem aquest nou índex per representar la posició de cada equip:

`0, 1, 2, 3...` → `1, 2, 3, 4...`

```python
def classificació_final(p_df_estadistiques_ordenat):

    classificacio_final = p_df_estadistiques_ordenat.reset_index()
    classificacio_final = classificacio_final[["equip", "punts"]]
    classificacio_final.index = classificacio_final.index + 1

    print("La classificació final és:")
    display(classificacio_final)


classificació_final(df_estadistiques_ordenat)
```

## 🧠 Algoritme principal

### ✔ `main()`

**Interpretació:** aquesta és la funció principal que coordina tot el procés de l'exercici. Va passant el resultat d'una funció a la següent fins a obtenir les estadístiques dels equips, ordenar la classificació i mostrar els resultats demanats.

1. Preparar les dades dels partits amb `preparar_dades()`.
2. Calcular les estadístiques de cada equip amb `estadistics()`.
3. Ordenar el DataFrame d'estadístiques per punts reutilitzant `ordenar_df()`.
4. Mostrar l'equip o equips més golejadors.
5. Mostrar l'equip o equips més golejats.
6. Mostrar el total de gols de cada equip.
7. Mostrar la classificació final amb els equips ordenats per punts.

**Flux mental:**

`df_partits` → `preparar_dades()` → `estadistics()` → `ordenar_df()` → mostrar resultats → `classificació_final()`

**Idea clau:** `main()` coordina tot el procés, mentre que cada subfunció s'encarrega d'una tasca concreta.

```python
def main_partits():
    
    df_partits_net = preparar_dades(df_partits)
    df_estadistiques = estadistics(df_partits_net)
    df_estadistiques_ordenat = ordenar_df(df_estadistiques, "punts", p_ascendent=False)
    equip_golejador(df_estadistiques)
    equip_golejat(df_estadistiques)
    total_gols(df_estadistiques)
    classificació_final(df_estadistiques_ordenat)


main_partits()
```
