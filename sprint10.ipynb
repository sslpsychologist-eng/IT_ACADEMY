# NIVELL 1

## 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

- Llista de publicacions (/posts)
- Llista d’usuaris (/users)
- Llista de tasques (/todos)

```python
url = "https://jsonplaceholder.typicode.com"

resposta_posts = requests.get(f"{url}/posts", timeout=10)
resposta_users = requests.get(f"{url}/users", timeout=10)
resposta_todos = requests.get(f"{url}/todos", timeout=10)
```

## 2. Mostra per pantalla:

- La quantitat total de cada recurs.
- El codi d'estat de cada petició.

```python
print(f"Posts: {len(resposta_posts.json())} registros | Estado HTTP: {resposta_posts.status_code}")
print(f"Users: {len(resposta_users.json())} registros | Estado HTTP: {resposta_users.status_code}")
print(f"Todos: {len(resposta_todos.json())} registros | Estado HTTP: {resposta_todos.status_code}")
```

## 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
nova_publi = {
    "title": "Exemple 1",
    "body": "Contingut de l'exemple 1",
    "userId": 1
}

resposta = requests.post(f"{url}/posts", json=nova_publi, timeout=10)

print(f"Resposta JSON: {resposta.json()} | Estat HTTP: {resposta.status_code}")
```

## 5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
canvi1 = {"title": "Títol actualitzat"}

resposta = requests.patch(f"{url}/posts/5", json=canvi1, timeout=10)

print(f"Resposta JSON: {resposta.json()} | Estat HTTP: {resposta.status_code}")
```

## 6. Fes una petició DELETE sobre una publicació.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
resposta = requests.delete(f"{url}/posts/5", timeout=10)

print(f"Resposta JSON: {resposta.json()} | Estat HTTP: {resposta.status_code}")
```

```python
url = "https://catfact.ninja"

resposta_facts = requests.get(f"{url}/facts", timeout=10)
resposta_breeds = requests.get(f"{url}/breeds", timeout=10)

print(resposta_facts.headers.get("Content-Type"))
print(resposta_breeds.headers.get("Content-Type"))
```

Paràmetres utilitzats a /facts:

Paràmetres utilitzats a /breeds:

## 3. Fes una petició GET senzilla:

- Mostra el codi d'estat de la resposta.
- Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
parametres = {"limit": 5}

resposta = requests.get(f"{url}/breeds", params=parametres, timeout=10)

print(f"Estat HTTP: {resposta.status_code}")

dades = resposta.json()
races = dades["data"]

print(f"Raça: {races[0]['breed']}")
print(f"País: {races[0]['country']}")
print(f"Origen: {races[0]['origin']}")
```

## 4. Converteix les dades a un DataFrame de pandas:

Mostra les primeres files del DataFrame.

```python
import pandas as pd

df_races = pd.DataFrame(races)

display(df_races.head())
```
