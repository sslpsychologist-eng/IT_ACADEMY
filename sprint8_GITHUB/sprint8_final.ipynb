### PAS 2. Codificació i accentuació

Un cop tenim l'estructura correcta, revisem si hi ha problemes de codificació en els noms de les columnes i en els valors de text.

#### Revisió de la codificació dels valors

Un cop corregits els noms de les columnes, es revisen les variables en què s'han detectat possibles problemes de codificació. S'hi inclouen principalment variables textuals i també `Salari mensual`, ja que, tot i que conceptualment hauria de ser una variable numèrica, actualment conté representacions textuals del símbol de l'euro amb errors de codificació.

Es crea la llista `columnes_codificacio` per agrupar únicament les columnes sobre les quals cal revisar aquest tipus de problema. Inicialment es visualitzen els seus valors únics mitjançant `unique()` per identificar els patrons incorrectes.


```python
columnes_codificacio = [
    "Nom",
    "Cognoms",
    "País d'origen",
    "Ciutat",
    "Gènere",
    "Càrrec",
    "Salari mensual"
]

for columna in columnes_codificacio:
    print(columna)
    print(df_complex[columna].unique())
```

#### Identificació dels valors amb possibles errors de codificació

La revisió dels valors únics mostra diferents patrons compatibles amb errors de codificació. En les variables de text apareixen seqüències que comencen per `Ã`, com en alguns noms i cognoms, mentre que a `Salari mensual` apareixen seqüències que comencen per `â`, associades a una representació incorrecta del símbol `€`.

Per facilitar-ne la identificació, es crea un filtre per a cada columna que selecciona els registres que contenen qualsevol dels dos patrons (`Ã` o `â`). El filtre genera valors booleans (`True` o `False`) i, mitjançant `loc`, es mostren només els valors afectats.


```python
for columna in columnes_codificacio:
    filtre = (
        df_complex[columna].astype(str).str.contains("Ã", na=False)
        |
        df_complex[columna].astype(str).str.contains("â", na=False)
    )
    print(columna)
    print(df_complex.loc[filtre, columna].unique())
```

#### Correcció dels errors de codificació

Després de comprovar diferents casos, s'observa que els valors afectats es poden corregir reinterpretant la seva codificació. La funció intenta primer recodificar el text amb `cp1252` i interpretar-lo com a `utf-8`. Com que alguns caràcters detectats no poden ser tractats amb `cp1252`, si aquesta opció falla es prova `latin1` com a alternativa.

La correcció només s'intenta quan el valor és de tipus text i conté algun dels patrons problemàtics identificats (`Ã` o `â`). Si cap de les dues conversions és possible, el valor es conserva sense modificar.

Finalment, la funció s'aplica amb `map()` a totes les columnes incloses a `columnes_codificacio`.


```python
def corregir_codificacio(p_text):

    if isinstance(p_text, str) and ("Ã" in p_text or "â" in p_text):

        try:
            return p_text.encode("cp1252").decode("utf-8")

        except UnicodeError:
            try:
                return p_text.encode("latin1").decode("utf-8")

            except UnicodeError:
                return p_text

    return p_text
```

```python
for columna in columnes_codificacio:
    df_complex[columna] = df_complex[columna].map(corregir_codificacio)
```

#### Comprovació de la correcció

Un cop aplicada la funció, es repeteix la cerca dels dos patrons utilitzats per detectar els errors (`Ã` i `â`). D'aquesta manera es comprova que la correcció s'ha aplicat correctament a totes les columnes revisades.


```python
for columna in columnes_codificacio:
    filtre = (
        df_complex[columna].astype(str).str.contains("Ã", na=False)
        |
        df_complex[columna].astype(str).str.contains("â", na=False)
    )
    print(columna, df_complex.loc[filtre, columna].unique())
```

#### Resultat de la comprovació

La comprovació final no detecta valors que continguin els patrons `Ã` o `â` en les columnes revisades. Per tant, es considera que els problemes de codificació identificats han estat corregits.

Aquesta correcció resol únicament els problemes de **codificació**. En el cas de `Salari mensual`, encara poden existir formats diferents, com imports amb o sense espais abans del símbol `€`, valors escrits amb la paraula `euros` o imports sense símbol. Aquestes diferències corresponen a problemes de **format i normalització**, no de codificació, i es tractaran posteriorment sense modificar-les en aquesta fase.

Després de revisar la distribució, els valors extrems poden influir molt en una mitjana. Per això, per imputar les temperatures absents s'utilitza la **mediana**, que és més robusta davant de valors extrems.

El criteri geogràfic és:

1. mediana de la mateixa **ciutat i país**;
2. si no hi ha prou informació, mediana del mateix **país**.

És important utilitzar `Ciutat` i `País d'origen` conjuntament, perquè poden existir ciutats amb el mateix nom en països diferents. Així evitem barrejar temperatures de llocs diferents.


```python
nuls_temperatura_abans = df_complex["Temperatura_mitjana_ciutat"].isna().sum()

mediana_ciutat_pais = df_complex.groupby(
    ["Ciutat", "País d'origen"]
)["Temperatura_mitjana_ciutat"].transform("median")

mediana_pais = df_complex.groupby(
    "País d'origen"
)["Temperatura_mitjana_ciutat"].transform("median")

df_complex["Temperatura_mitjana_ciutat"] = (
    df_complex["Temperatura_mitjana_ciutat"]
    .fillna(mediana_ciutat_pais)
    .fillna(mediana_pais)
)

nuls_temperatura_despres = df_complex["Temperatura_mitjana_ciutat"].isna().sum()

print(f"Temperatures absents abans: {nuls_temperatura_abans}")
print(f"Temperatures absents després: {nuls_temperatura_despres}")
```

### 3.5. Preparació i imputació del salari mensual

Abans de netejar el salari, s'observa la columna original per identificar com estan representats realment els imports.

```python
print(f"Tipus original: {df_complex['Salari mensual'].dtype}")
print(
    "Valors nuls originals:",
    df_complex["Salari mensual"].isna().sum()
)

display(
    df_complex[["Salari mensual"]]
    .drop_duplicates()
    .head(30)
)
```

```python
display(df_complex["Salari mensual"].map(type).value_counts())
```

La columna `Salari mensual` conté diferents formats: símbol `€`, text (`euros`), separadors de milers, imports escrits amb paraules i el valor `n/d`.

Primer es normalitzen aquests formats i després es converteix la columna a numèrica. Els valors `n/d` passen a ser nuls perquè no contenen un salari recuperable.


```python
mapa_salaris = {
    "mil vuit-cents": "1800",
    "dos mil": "2000",
    "tres mil": "3000",
    "n/d": pd.NA
}

salari_net = (
    df_complex["Salari mensual"]
    .str.lower()
    .str.strip()
    .replace(mapa_salaris)
    .str.replace("euros", "", regex=False)
    .str.replace("euro", "", regex=False)
    .str.replace("€", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace(".", "", regex=False)
    .str.replace(" ", "", regex=False)
)

salari_numeric = pd.to_numeric(salari_net, errors="coerce")

print(f"Valors que queden nuls: {salari_numeric.isna().sum()}")
```

Un cop normalitzat el format, es substitueix la columna original per la versió numèrica. Els valors que no contenien un salari recuperable queden com a `NaN` i s'imputaran a continuació.


```python
df_complex["Salari mensual"] = salari_numeric

nuls_salari_abans = df_complex["Salari mensual"].isna().sum()

display(df_complex["Salari mensual"].describe())

print(f"Tipus de dada del salari: {df_complex['Salari mensual'].dtype}")
print(f"Salaris absents abans de la imputació: {nuls_salari_abans}")
```

Per imputar els salaris que falten s'utilitza la **mediana** perquè els salaris poden tenir valors molt alts o molt baixos i aquests extrems afecten més la mitjana.

El criteri és:

1. mediana del mateix `Grup Professional` i `Càrrec`;
2. si no és possible, mediana del mateix `Grup Professional`;
3. com a últim recurs, mediana global.

Així es prioritza la informació laboral més semblant a la persona i s'evita que salaris extrems distorsionin massa el valor imputat.


```python
mediana_grup_carrec = df_complex.groupby(
    ["Grup Professional", "Càrrec"]
)["Salari mensual"].transform("median")

mediana_grup = df_complex.groupby(
    "Grup Professional"
)["Salari mensual"].transform("median")

df_complex["Salari mensual"] = (
    df_complex["Salari mensual"]
    .fillna(mediana_grup_carrec)
    .fillna(mediana_grup)
    .fillna(df_complex["Salari mensual"].median())
)

print("Salaris absents després de la imputació:", df_complex["Salari mensual"].isna().sum())
```

#### Conclusió del salari

Primer s'han normalitzat els formats recuperables i després s'han imputat només els salaris que faltaven.

S'utilitza la **mediana** perquè és més robusta davant de salaris extrems. Primer es busca un valor representatiu dins del mateix grup professional i càrrec, després dins del grup professional i, només si encara falta informació, s'utilitza la mediana global.

#### Validació de la data de naixement

### Càlcul i categorització de l'edat

Amb les dates vàlides es calcula l'edat actual tenint en compte si la persona ja ha complert anys aquest any. Posteriorment, `pd.cut()` transforma l'edat en les mateixes franges definides per la política salarial: `18-29`, `30-44`, `45-59` i `60+`.

Els registres amb una data de naixement no vàlida mantenen l'edat i la categoria com a valors nuls, de manera que no se'ls atribueix un ajust d'edat inventat.


```python
avui = pd.Timestamp.today()

df_enriquit["Edat"] = (
    avui.year - df_enriquit["Data_naixement"].dt.year
    - (
        (df_enriquit["Data_naixement"].dt.month > avui.month)
        | (
            (df_enriquit["Data_naixement"].dt.month == avui.month)
            & (df_enriquit["Data_naixement"].dt.day > avui.day)
        )
    )
)
```

```python
df_enriquit["Categoria_edat"] = pd.cut(
    df_enriquit["Edat"],
    bins=[17, 29, 44, 59, float("inf")],
    labels=["18-29", "30-44", "45-59", "60+"]
)
```

### Assignació dels ajustos de la política salarial

Un cop preparades totes les categories necessàries, s'utilitza `.map()` per relacionar cada categoria amb el percentatge definit al JSON. Es creen per separat els ajustos de grup professional, fills, edat, cost de vida i transport.

Mantenir els ajustos en columnes independents permet comprovar l'origen de l'increment final i detectar fàcilment els casos en què algun ajust no es pot calcular.


```python
df_enriquit["Increment_grup"] = df_enriquit["Grup Professional"].map(politica["base_increment_by_group"])

df_enriquit["Increment_fills"] = df_enriquit["Categoria_fills"].map(politica["children_adjustment"])

df_enriquit["Increment_edat"] = df_enriquit["Categoria_edat"].map(politica["age_adjustment"])

df_enriquit["Increment_cost_vida"] = df_enriquit["Categoria_cost_vida"].map(politica["cost_of_living_adjustment"])

df_enriquit["Increment_transport"] = df_enriquit["Categoria_transport"].map(politica["transport_adjustment"])
```

```python
display(
    df_enriquit[
        ["Grup Professional", "Increment_grup",
         "Categoria_fills", "Increment_fills",
         "Categoria_edat", "Increment_edat",
         "Categoria_cost_vida", "Increment_cost_vida",
         "Categoria_transport", "Increment_transport"]
    ].head(20)
)
```

### Càlcul de l'increment salarial total

Els cinc ajustos es sumen fila per fila per obtenir `Increment_total`. Quan falta informació per a algun ajust, el valor `NaN` s'ignora i es calcula l'increment únicament amb la informació disponible, d'acord amb la decisió de conservar els registres incomplets.

A continuació s'aplica el límit màxim definit al JSON, de manera que cap increment total pugui superar el 15%.


```python
columnes_increment = [
    "Increment_grup",
    "Increment_fills",
    "Increment_edat",
    "Increment_cost_vida",
    "Increment_transport"
]

df_enriquit["Increment_total"] = df_enriquit[columnes_increment].sum(
    axis=1,
    skipna=True
)
```

```python
increment_maxim = politica["rules"]["max_total_increment_percent"]

df_enriquit["Increment_total"] = df_enriquit["Increment_total"].clip(
    upper=increment_maxim
)
```

### Càlcul del salari ajustat

El salari ajustat es calcula aplicant el percentatge d'increment total al salari mensual original. El resultat s'arrodoneix a dos decimals per expressar-lo en format monetari.


```python
df_enriquit["Salari_ajustat"] = (
    df_enriquit["Salari mensual"]
    * (1 + df_enriquit["Increment_total"] / 100)
).round(2)
```

### Funció d'exploració visual automatitzada

Per automatitzar l'exploració inicial d'un DataFrame, es crea una funció que recorre totes les seves columnes i genera un gràfic adaptat al tipus de variable.

La funció segueix aquesta lògica:

- Si la variable és **numèrica**, es genera un histograma per observar-ne la distribució.
- Si la variable és de tipus **data**, s'extreu l'any i es representa la freqüència de registres per any.
- En la resta de casos, la variable es tracta com a **categòrica** i es representen els 10 valors més freqüents mitjançant un gràfic de barres.

Els valors nuls de les variables numèriques s'exclouen únicament per generar el gràfic mitjançant `dropna()`, sense modificar el DataFrame original.

Cada gràfic es guarda automàticament en format `.png`, utilitzant el nom de la columna com a nom del fitxer. Finalment, la figura es tanca amb `plt.close()` per evitar acumular figures obertes durant l'execució del bucle.

```python
def grafics(p_df):

    for columna in p_df.columns:

        nom_fitxer = columna.replace(" ", "_")

        # Variables numèriques
        if pd.api.types.is_numeric_dtype(p_df[columna]):

            plt.figure(figsize=(8, 5))

            plt.hist(
                p_df[columna].dropna(),
                bins=20
            )

            plt.title(f"Distribució de {columna}")
            plt.xlabel(columna)
            plt.ylabel("Freqüència")

            plt.savefig(f"{nom_fitxer}.png")
            plt.show()
            plt.close()

        # Variables de tipus data
        elif pd.api.types.is_datetime64_any_dtype(p_df[columna]):

            anys = (
                p_df[columna]
                .dt.year
                .value_counts()
                .sort_index()
            )

            plt.figure(figsize=(8, 5))

            anys.plot(kind="bar")

            plt.title(f"Distribució per any de {columna}")
            plt.xlabel("Any")
            plt.ylabel("Freqüència")

            plt.savefig(f"{nom_fitxer}.png")
            plt.show()
            plt.close()

        # Variables categòriques
        else:

            freq = (
                p_df[columna]
                .value_counts()
                .head(10)
            )

            plt.figure(figsize=(8, 5))

            freq.plot(kind="bar")

            plt.title(f"Valors més freqüents de {columna}")
            plt.xlabel(columna)
            plt.ylabel("Freqüència")

            plt.savefig(f"{nom_fitxer}.png")
            plt.show()
            plt.close()
```

## Exercici 2: Optimització de rutes (problema aplicat)

- Rebi: una ciutat d'origen i una ciutat de destinació.
- Retorni: l'ordre de les ciutats visitades (ruta) i la distància total recorreguda.

La ruta s'ha de construir seguint una estratègia senzilla: anar sempre a la ciutat no visitada més propera fins a arribar a la destinació.

### Preparació de la matriu de distàncies

Es carrega el fitxer `matriu_distancies.xlsx` utilitzant la primera columna com a índex, ja que tant les files com les columnes representen ciutats i els valors de la matriu indiquen la distància en quilòmetres entre elles.

Seguint les indicacions de l'enunciat, s'eliminen `Palma` i `Las Palmas de Gran Canaria` tant de les files com de les columnes, mantenint així una matriu quadrada amb les mateixes ciutats als dos eixos.

```python
ciutats_eliminar = ["Palma", "Las Palmas de Gran Canaria"]

df_distancies = df_distancies.drop(
    index=ciutats_eliminar,
    columns=ciutats_eliminar
)
```

### Funció per calcular la ruta

La funció rep una ciutat d'origen i una ciutat de destinació.

Es crea una llista `ruta` amb la ciutat inicial, una variable `ciutat_actual` i un acumulador `distancia_total`.

Mitjançant un bucle `while`, mentre no s'arriba a la destinació:

- Es consulten les distàncies des de la ciutat actual amb `loc`.
- S'eliminen les ciutats ja visitades i els valors nuls.
- Amb `idxmin()` es selecciona la ciutat no visitada més propera.
- S'afegeix la distància al total acumulat.
- S'afegeix la nova ciutat a la ruta.
- La nova ciutat passa a ser la ciutat actual.

La funció retorna la ruta seguida i la distància total recorreguda.

```python
def calcular_ruta(p_origen, p_destinacio):

    ruta = [p_origen]
    ciutat_actual = p_origen
    distancia_total = 0

    while ciutat_actual != p_destinacio:

        distancies = df_distancies.loc[ciutat_actual]
        distancies_no_visitades = distancies.drop(labels=ruta).dropna()        
        seguent_ciutat = distancies_no_visitades.idxmin()        
        distancia_pas = distancies_no_visitades.loc[seguent_ciutat]        
        distancia_total += distancia_pas        
        ruta.append(seguent_ciutat)        
        ciutat_actual = seguent_ciutat

    return ruta, distancia_total
```

```python
ruta, distancia_total = calcular_ruta("Barcelona", "Valladolid")

print("Ruta:", ruta)
print("Distància total:", distancia_total, "km")
```

### Resultat

La funció retorna l'ordre de les ciutats visitades des de l'origen fins a la destinació i la distància total recorreguda.

La ruta es construeix seguint una estratègia `greedy`, escollint en cada pas la ciutat no visitada més propera. Per aquest motiu, el resultat és una aproximació i no garanteix necessàriament la ruta globalment més curta.
